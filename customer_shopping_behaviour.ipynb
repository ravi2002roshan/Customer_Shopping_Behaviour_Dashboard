Checking if any null values present in the column or not and there is null value present in "Review Rating" column

```python
df.isnull().sum()
```

Here we are handling the Null values present in the Review Rating column with Median

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

Proper naming of the columns by changing them to lower case and adding hyphen in between them

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.columns
```

Changing the name of the "purchase_amount_(USD)" to "purchase_amount"

```python
df = df.rename(columns = { 'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

Doing some Feature Enginering and creating a column

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

Checking the created column

```python
df[['age','age_group']].head(10)
```

Creating column "purchase_frequency_days"

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

Checking if "discount_applied" is directly related to "promo_code_applied" or vice-versa
